```python
import pyspark 
from pyspark.sql import SparkSession # this gives us opp to create SC
```

```python
# Spark context is the entry for your spark application,
# we need to create a spark context 
# for this code 
sc = SparkSession.builder.master("local")\
        .appName("djsbalakrishnan-demo")\
        .getOrCreate()\
        .sparkContext
```

```python
# how to fetch the data from HDFS 
movies_rdd = sc.textFile("/user/userjuly2025019/movies.csv")
# abstraction on top of my file which is in HDFS 
```

```python
movies_rdd.collect()
```

```python
movies_rdd.count()
```

```python
## RDD creation using parallelized collection

l = list(range(100)) # python list
print(l)
```

```python
list_rdd = sc.parallelize(l, 10)
list_rdd.collect()
```

```python
## Lazy evaluation 
movies_demo_rdd = sc.textFile("/user/userjuly2025019/spark/dhiraj.csv") # no action 
```

```python
movies_demo_rdd.collect()

# Only when an action is performed, the file will be read
# If the action is not performed, the DAG is created, without any action taken 
```

```python
## To Debug String explanation 

debug_list = list(range(100))
```

```python
debug_rdd = sc.parallelize(debug_list)

debug_rdd.take(5)
```

```python
# I want to add + 5 to each of my element 

# map () -> iterator
def addFive(element):
    return element + 5

modified_rdd = debug_rdd.map(addFive)
modified_rdd.take(5)
```
